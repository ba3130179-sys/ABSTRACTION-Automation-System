```python
from abc import ABC, abstractmethod
class Appliance(ABC):
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
        self.is_on = False

    @abstractmethod
    def turn_on(self):
        """Must be implemented by every child class"""
        pass

    @abstractmethod
    def turn_off(self):
        """Must be implemented by every child class"""
        pass

    def get_status(self):
        state = "ON" if self.is_on else "OFF"
        return f"{self.brand} {self.model} is {state}"


class WashingMachine(Appliance):
    def turn_on(self):
        self.is_on = True
        return f"WashingMachine {self.brand}: Filling drum with water. Starting wash cycle."

    def turn_off(self):
        self.is_on = False
        return f"WashingMachine {self.brand}: Draining water. Cycle complete."


class Refrigerator(Appliance):
    def turn_on(self):
        self.is_on = True
        return f"Refrigerator {self.brand}: Compressor started. Cooling to 4°C."

    def turn_off(self):
        self.is_on = False
        return f"Refrigerator {self.brand}: Compressor stopped. Defrost mode activated."


class Microwave(Appliance):
    def turn_on(self):
        self.is_on = True
        return f"Microwave {self.brand}: Magnetron activated. Ready to heat."

    def turn_off(self):
        self.is_on = False
        return f"Microwave {self.brand}: Magnetron off. Door unlocked."
print("APPLIANCE DEMO ")
appliances = [
    WashingMachine("Samsung", "EcoBubble"),
    Refrigerator("LG", "SmartInverter"),
    Microwave("Panasonic", "NN-SN966S")
]

for appliance in appliances:
    print(f"\n{appliance.get_status()}")
    print(appliance.turn_on())
    print(appliance.get_status())
    print(appliance.turn_off())
    print(appliance.get_status())
```
